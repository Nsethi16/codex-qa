# PDF to Base64

Use the widget below to select a local PDF file. The notebook will print the Base64-encoded content.

```python
import base64
from IPython.display import display
import ipywidgets as widgets

uploader = widgets.FileUpload(accept='.pdf', multiple=False)
output = widgets.Output()

def handle_upload(change):
    output.clear_output()
    if not uploader.value:
        return
    upload = next(iter(uploader.value.values()))
    encoded = base64.b64encode(upload['content']).decode('utf-8')
    with output:
        print(encoded)

uploader.observe(handle_upload, names='value')
display(uploader, output)
```

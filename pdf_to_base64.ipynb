# PDF to Base64

Provide the local path to a PDF file in the input cell below. The notebook will print the Base64-encoded content.

```python
import base64
from pathlib import Path

pdf_path = Path('path/to/file.pdf')

if not pdf_path.is_file():
    raise FileNotFoundError(f'PDF not found: {pdf_path.resolve()}')

encoded = base64.b64encode(pdf_path.read_bytes()).decode('utf-8')
print(encoded)
```
